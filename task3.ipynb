```python
import pandas as pd
import numpy as np
```

```python
import seaborn as sns

```

```python
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\Datascience_internship.zip\dataset\Dataset .csv")
```

```python
plt.hist(df['Aggregate rating'], bins=20, edgecolor='black', color='skyblue')
```

```python
average_rating = df.groupby('Cuisines')['Aggregate rating'].mean().sort_values(ascending=False).head(10)
```

```python
plt.barh(average_rating.index, average_rating.values, color='lightgreen')
plt.xlabel('Average Aggregate Rating')
plt.ylabel('Cuisines')
plt.title('Top 10 Cuisines by Average Aggregate Rating')
plt.show()
```

```python
city_rating = df.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False).head(10)
```

```python
plt.barh(city_rating.index, city_rating.values, color='salmon')
plt.xlabel('Average Aggregate Rating')
plt.ylabel('Cities')
plt.title('Top 10 Cities by Average Aggregate Rating')
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df['Votes'],
    df['Aggregate rating'],
    alpha=0.5
)

plt.xlabel('Number of Votes')
plt.ylabel('Aggregate Rating')
plt.title('Relationship Between Votes and Restaurant Rating')

plt.show()
```
